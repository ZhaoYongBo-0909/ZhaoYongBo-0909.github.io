#  fileInput 函数：fileInput


Original link ：https://rdrr.io/cran/shiny/man/fileInput.html

```r
# 官网实例

ui <- fluidPage(
  sidebarLayout(
    sidebarPanel(     # 定义侧边栏面板
      fileInput("file1", "Choose CSV File", accept = ".csv"),
      
      # 添加文件输入控件，允许用户上传文件  
      # fileInput的第一个参数是输入控件的ID（"file1"），用于在server函数中引用  
      # 第二个参数是标签（"Choose CSV File"），显示在网页上  
      # accept参数限制只能上传.csv文件  
    
      checkboxInput("header", "Header", TRUE)
            
      # 添加复选框控件，用于指定CSV文件是否包含表头  
      # checkboxInput的第一个参数是控件的ID（"header"），第二个参数是标签（"Header"），第三个参数是默认值（TRUE）
    ),
  # 定义主面板  
    mainPanel(  
      # 在主面板中添加表格输出控件，用于显示数据  
      # tableOutput的第一个参数是输出控件的ID（"contents"），用于在server函数中引用  
      tableOutput("contents")  

    
    )
  )
)

# 定义服务器逻辑（Server）  
server <- function(input, output) {  
  # 使用renderTable函数渲染表格，并将其输出到UI中ID为"contents"的控件
  output$contents <- renderTable({  
    # 从输入控件中获取文件对象  
    file <- input$file1  
    # 使用tools包中的file_ext函数获取文件扩展名  
    # 注意：这里假设file$datapath已经是一个有效的文件路径，这在用户上传文件后通常是成立的
    ext <- tools::file_ext(file$datapath)  
      
    # 使用req函数检查文件是否已被选择  
    # 如果file为NULL（即用户没有选择文件），则req会停止执行并显示错误消息
    req(file)  
      
    # 使用validate和need函数进行输入验证
    # 确保文件扩展名为".csv"，如果不是，则显示错误消息"Please upload a csv file"
    validate(need(ext == "csv", "Please upload a csv file"))
      
    # 读取CSV文件，header参数根据复选框控件的值来确定是否包含表头
    # file$datapath是上传文件的路径，read.csv函数读取该文件并返回数据框（data.frame）
    read.csv(file$datapath, header = input$header)
  })  
}  

shinyApp(ui, server)
```
